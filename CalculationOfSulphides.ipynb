```python
import pandas as pd
import os
os.chdir("/content/drive/My Drive/group research/Geophysics Group/New Database/Petrophysics Data/Lisheen/campaign04/")
xrf = pd.read_excel("campaign04_integrated_database.xlsx", sheet_name='xrf')
new_xrf= xrf[["sample_id", "Zn Mean", "Pb Mean", "Fe Mean"]]
ZnS = xrf["Zn Mean"]*0.0001/0.6709
PbS = xrf["Pb Mean"]*0.0001/0.866
FeS2= xrf["Fe Mean"]*0.0001/0.4655
new_xrf['ZnS%'] = ZnS
new_xrf['PbS%'] = PbS
new_xrf['FeS2%'] = FeS2
new_xrf['TotalSul%'] = new_xrf['ZnS%']+new_xrf['PbS%']+new_xrf['FeS2%']

new_xrf['isMS?'] = ['Ms' if x > 40 else '0' for x in new_xrf['TotalSul%'] ]
pd.set_option('display.max_rows', None)
new_xrf
```
